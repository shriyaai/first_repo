```python
class Circle:
  def __init__(self,radius):
    self.radius = radius
  def getArea(self):
    return 3.14*self.radius*self.radius
  def getCircumference(self):
    return self.radius*2*3.14
a = Circle(6)
print(a.getArea())
print(a.getCircumference())
```

```python
class Temperature:
  def __init__(self,temt):
    self.temt = temt
  def convertFarenhit(self):
    return (self.temt*9/5)+32
  def convertCelcius(self):
    return (self.temt-32)*5/9
temt1=Temperature(35)
print(temt1.convertCelcius())
print(temt1.convertFarenhit())
```

```python
class Time:
    def __init__(self, h, m):
        self.h = h
        self.m = m

    def addTime(self, t1, t2):
        self.h = t1.h + t2.h
        self.m = t1.m + t2.m
        if self.m >= 60:
            self.h += 1
            self.m -= 60

    def displayTime(self):
        print(self.h, "hour", self.m, "minute")

    def displayMinute(self):
        print(self.h * 60 + self.m)


t1 = Time(2, 50)
t2 = Time(1, 20)

t3 = Time(0, 0)
t3.addTime(t1, t2)
t3.displayTime()

t4 = Time(1, 2)
t4.displayMinute()

```
